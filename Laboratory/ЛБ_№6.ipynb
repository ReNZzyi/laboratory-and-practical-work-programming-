```python
!wget -O pr6_1.txt https://raw.githubusercontent.com/ReNZzyi/laboratory-and-practical-work-programming-/main/Data/pr6_1.txt

with open('pr6_1.txt') as f:
    s = f.readline().strip()

count = 0
for i in range(len(s) - 1):
    if s[i].isdigit() and s[i + 1].isdigit():
        count += 1

print(count)
```
