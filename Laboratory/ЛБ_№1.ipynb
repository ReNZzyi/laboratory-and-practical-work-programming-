## Задание

Необходимо вычислить значение выражения:
$$
t = \frac{35}{a} \cdot z + z \cdot a - \frac{a + E_t}{4}
$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $z, a, E_t$. Переменная, которая должна быть вычислена на основе независимых: $t$.

```python
z = 37
a = 22.5
E_t = 30.03

print("z = ", z, ", type(z) = ", type(z))
print("a = ", a, ", type(a) = ", type(a))
print("E_t = ", E_t, ", type(E_t) = ", type(E_t))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $t$:

```python
t = (35 / a * z + z * a - (a + E_t) / 4)
print('t =',t)
print('Type of t: ', type(t))
```
