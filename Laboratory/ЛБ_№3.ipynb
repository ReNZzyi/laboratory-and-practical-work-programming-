## Задание


Напишите функцию, которая по трем переданным ей числам
определяет могут ли эти числа быть сторонами одного и того же
остроугольного треугольника. Функция возвращает True или False

## Решение

### Объявление функции

```python
def is_acute_triangle(a, b, c):
    if a <= 0 or b <= 0 or c <= 0:
        return False
    if a + b <= c or a + c <= b or b + c <= a:
        return False

    sides = sorted([a, b, c])
    x, y, z = sides

    return x**2 + y**2 > z**2
```

### Демонстрация использования функции

```python
print(is_acute_triangle(3, 4, 5))
print(is_acute_triangle(4, 5, 6))
print(is_acute_triangle(2, 2, 2))
print(is_acute_triangle(1, 1, 10))
print(is_acute_triangle(5, 6, 10))
```
