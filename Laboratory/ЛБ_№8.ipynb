## Задание

Григорий придумывает 16-буквенные слова, состоящие из букв слова АНТИУТОПИЯ. Сколько слов, содержащих комбинацию АНТИУТОПИЯ, может составить Григорий, если количество гласных справа от этой комбинации отличается от количества гласных слева на 1? Буквы в словах могут повторяться любое количество раз или же не встречаться вовсе.

## Решение

Слово АНТИУТОПИЯ состоит из 8 различных букв: гласные А, И, У, О, Я и согласные Н, Т, П. Комбинация занимает 10 букв, значит, остаются 6 свободных позиций: `L` букв слева и `6 - L` справа (`L` от 0 до 6).

Перебор всех 16-буквенных слов (8<sup>16</sup> вариантов) слишком долгий, поэтому переберём только 6 свободных букв функцией `itertools.product` (8<sup>6</sup> вариантов) и все способы разделить их на левую и правую части. Для каждого варианта сравним количество гласных слева и справа.

Каждое слово при этом посчитается ровно один раз: комбинация не может встретиться в слове дважды, так как два её вхождения не помещаются в 16 букв и не могут перекрываться.

```python
import itertools

alf = 'АНТИУОПЯ'   # различные буквы слова АНТИУТОПИЯ
glas = 'АИУОЯ'     # гласные
count = 0

for p in itertools.product(alf, repeat=6):
  for L in range(7):          # L букв слева, 6 - L справа
    left = p[:L]
    right = p[L:]
    gl_left = sum(1 for ch in left if ch in glas)
    gl_right = sum(1 for ch in right if ch in glas)
    if abs(gl_right - gl_left) == 1:
      count += 1

print(count)
```
