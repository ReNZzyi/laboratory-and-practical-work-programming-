## Решение задачи 1

```python
s = open('data_v1.txt').read()

count = {}
for c in s:
    if c.isupper():
        count[c] = count.get(c, 0) + 1

m = max(count.values())
for c in count:
    if count[c] == m:
        print(c, m)
```

## Решение задачи 2

```python
s = open('data_v1.txt').read()

letters = set(c for c in s if c.isupper())

m = 0
for c in letters:
    if s.count(c) > m:
        m = s.count(c)

for c in letters:
    if s.count(c) == m:
        print(c, m)
```

## Решение задачи 3

```python
from collections import Counter

s = open('data_v1.txt').read()

count = Counter(c for c in s if c.isupper())

m = max(count.values())
for c in count:
    if count[c] == m:
        print(c, m)
```
