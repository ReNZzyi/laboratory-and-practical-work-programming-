## Решение

```python
   print(open('pr5.txt').read()[:50])

   a = []
   for line in open('pr5.txt'):
       if line.strip() != '':
           a.append(int(line))

   sr = sum(a) / len(a)

   k = 0
   mx = -100000
   for i in range(len(a) - 2):
       x = a[i]
       y = a[i + 1]
       z = a[i + 2]
       cnt = 0
       if x > sr:
           cnt = cnt + 1
       if y > sr:
           cnt = cnt + 1
       if z > sr:
           cnt = cnt + 1
       if cnt >= 2:
           k = k + 1
           if x + y + z > mx:
               mx = x + y + z

   print(k, mx)
```
