## 1. Индивидуальное задание

Вычислить значение выражения:

$$Z=\frac{Y^{2}+X}{1\cdot 2}-\frac{Y^{2}-X^{4}}{2\cdot 4}+\frac{X^{4}+Y^{6}}{4\cdot 6}-\frac{X^{8}-Y^{6}}{6\cdot 8}+\dots$$

**Закономерность.** Множители в знаменателе - это пары `a, b`, где каждая следующая пара получается так: `a = b`, `b = b + 2` (первая пара `1, 2`). Числитель и знак каждого слагаемого повторяются с периодом 4.

```python
X = 7
Y = 3.5
n = 5

Z = 0
sign = 1
a, b = 1, 2

for k in range(1, n + 1):
    if k % 4 == 1:
        num = Y**b + X**a
    elif k % 4 == 2:
        num = Y**a - X**b
    elif k % 4 == 3:
        num = X**a + Y**b
    else:
        num = X**b - Y**a

    Z += sign * num / (a * b)

    sign = -sign
    a, b = b, b + 2

print("Тип X:", type(X))
print("Тип Y:", type(Y))
print("Тип n:", type(n))
print("Тип Z:", type(Z))
print("Z =", Z)
```
